# Run the sustainability analysis in Google Colab

Choose **Runtime → Run all**. The first cell clones the author’s main branch, installs its pinned packages, and runs the analysis. Results are saved in the Colab session under `replication_results/`.

This notebook clones the author's repository, installs the pinned dependencies, and runs the same command-line analysis script as the local instructions.

```python
from pathlib import Path
import subprocess
import sys

REPOSITORY = 'https://github.com/Sadullah4535/Global-Sustainability-Performance-.git'
BRANCH = 'main'
PROJECT_DIR = Path('/content/Global-Sustainability-Performance')

subprocess.run(
    ['git', 'clone', '--depth', '1', '--branch', BRANCH, REPOSITORY, str(PROJECT_DIR)],
    check=True,
)
subprocess.run(
    [sys.executable, '-m', 'pip', 'install', '-r', str(PROJECT_DIR / 'requirements.txt')],
    check=True,
)
subprocess.run(
    [sys.executable, str(PROJECT_DIR / 'run_replication.py')],
    cwd=PROJECT_DIR,
    check=True,
)
print(f'Results are in {PROJECT_DIR / "replication_results"}')
```
