# Run the sustainability analysis in Google Colab

Choose **Runtime → Run all**. The first cell clones the reproducibility branch, installs its pinned packages, and runs the analysis. Results are saved in the Colab session under `replication_results/`.

The launcher currently targets Manh's fork and `reproducibility-pipeline` branch so the workflow can be tried while the contribution is under review. After it is merged, the repository URL and branch can be switched to the author's `main`.

```python
from pathlib import Path
import subprocess
import sys

REPOSITORY = 'https://github.com/qmanhbeo/Global-Sustainability-Performance.git'
BRANCH = 'reproducibility-pipeline'
PROJECT_DIR = Path('/content/Global-Sustainability-Performance')

subprocess.run(
    ['git', 'clone', '--depth', '1', '--branch', BRANCH, REPOSITORY, str(PROJECT_DIR)],
    check=True,
)
subprocess.run(
    [sys.executable, '-m', 'pip', 'install', '-r', str(PROJECT_DIR / 'requirements.txt')],
    check=True,
)
subprocess.run(
    [sys.executable, str(PROJECT_DIR / 'run_replication.py')],
    cwd=PROJECT_DIR,
    check=True,
)
print(f'Results are in {PROJECT_DIR / "replication_results"}')
```
